### Three ideas you must keep separate

In the probability part, we will often compare three things. Keep them clear in your mind.

1. **Theoretical probability distribution:** a **math model**. We *assume* a rule (for example, "each region is equally likely") and calculate probabilities from the formula. No data is needed.
2. **Observed (actual) distribution:** what we **really see** in the dataset. We count values and calculate proportions.
3. **Checking the model:** we **compare** 1 and 2. If they are close, the model is a reasonable description of the data. If they are far apart, the model is not a good fit.

> ⚠️ We will **not force** the data to fit a distribution. Sometimes the honest answer is: *"This model does not fit this data well."* That is a good and useful answer.

**Defining a random variable from data.** Some distributions do not appear directly in the data. Then we **create** a variable. For example, `smoker` is "yes" or "no". We can write it as **1 = success (smoker)** and **0 = failure (non-smoker)**. Now we can study it with the Bernoulli distribution. Here "success" only means "the event we are counting". It does not mean something good.

### Exercise 13: Addition Rule of Probability

**1. Concept**

The **addition rule** gives the probability that **A or B** happens:

**P(A or B) = P(A) + P(B) − P(A and B)**

We subtract P(A and B) because those outcomes were counted twice.

If A and B **cannot happen together** (they are **mutually exclusive**), then P(A and B) = 0 and **P(A or B) = P(A) + P(B)**.

**2. Why Do We Need It?**

We often need the chance that **at least one** of two things happens. The addition rule stops us from double-counting.

**3. Real-World Example**

## Section 9: Distribution Analysis — Part E: Distribution Characteristics

So far we have looked at the **centre** and **spread** of data. Now we look at its **shape**.

- **Skewness:** is the data balanced, or does it have a long tail on one side?
- **Kurtosis:** does the data have heavy tails (many extreme values) or light tails (few extreme values)?
- **Z-score:** how far is one value from the mean, measured in standard deviations?
- **Empirical Rule:** for Normal data, how many values lie within 1, 2 and 3 standard deviations of the mean?

### Why does real data often NOT follow a Normal distribution?

The Normal curve is a perfect, symmetric, mathematical shape. Real data is often different because:
- **Natural limits:** some values cannot go below 0 (money, time, weight), but they can go very high. This creates a long right tail.
- **Mixed groups:** data may combine different groups (smokers and non-smokers), each with its own centre. This can create two humps.
- **Human choices and rules:** people choose how many children to have; ages in this dataset stop at 64 because of how the insurance works.
- **Rare big events:** a serious illness can create a very large bill.

So we should always **check** the shape, not just assume it is Normal.

## Where Do We Use These Statistical Concepts in AI/ML?

Everything you practised in this assignment is used in AI and Machine Learning. Here is how, in simple words.

| Statistics Concept | Where It Is Used in AI/ML | Simple example |
|---|---|---|
| Mean and Median | Filling missing values in data | If some ages are missing, fill them with the median age. Use the median when data is skewed (like charges), because the mean would be pulled by extreme values. |
| Mode | Filling missing categorical values | If some `region` values are missing, fill them with the most common region (southeast). |
| Quartiles and IQR | Finding unusual values (outliers) | Before training a model, use the 1.5 × IQR rule to find unusual charges, then decide whether they are errors or real cases. |
| Variance and Standard Deviation | Understanding data spread and feature scaling | Age is 18–64 but charges are up to 63,770. Many models work better when features are scaled using the mean and SD, so no feature dominates just because its numbers are bigger. A feature with almost zero variance carries little information. |
| Histograms | Understanding feature distributions | A histogram of charges shows strong right skew, so an analyst may transform it (for example, with a log) before modelling. |
| Probability | Classification and prediction | A model says: "There is a 0.83 probability this customer will have high charges." |
| Bayes' Theorem | Naive Bayes classification | A spam filter starts with P(spam) and updates it using the words in the email to get P(spam given the words), just like P(smoker given a high bill). |
| Random Variables | Modelling uncertain outcomes | The future medical cost of a new customer is a random variable. A model tries to describe and predict it. |
| PMF and PDF | Understanding probability distributions | A classifier outputs a PMF over classes (e.g. 70% cat, 20% dog, 10% bird). Some models assume features follow a PDF such as the Normal. |
| CDF | Finding probabilities below a particular value | "What is the probability that this customer's cost will be below $10,000?" Also used to choose thresholds and percentiles. |
| Expected Value | Prediction and decision-making | A regression model predicts the expected charge for a person. A business chooses the action with the highest expected profit. |
| Bernoulli Distribution | Modelling binary outcomes | Logistic regression models a yes/no target (smoker or not, spam or not) as Bernoulli with a predicted p. |
| Binomial Distribution | Modelling success counts across repeated trials | If a model is 80% accurate, the number of correct predictions out of 50 test cases can be modelled as Binomial(50, 0.8). |
| Poisson Distribution | Predicting event counts | Predicting the number of insurance claims per day, or website visits per hour. |
| Discrete Uniform Distribution | Modelling equally likely outcomes | Randomly shuffling data or picking a random sample for training and testing, where every row is equally likely to be chosen. |
| Normal Distribution | Statistical modelling and standardization | Many methods assume errors are roughly Normal. Standardized features (Z-scores) are centred at 0 with SD 1. |
| Skewness and Kurtosis | Understanding data distribution and unusual patterns | High skewness (charges) suggests a transformation. High kurtosis warns that extreme values may affect the model. |
| Z-score | Standardization and anomaly detection | A bank flags a card payment with z = 5 compared with the customer's usual spending as possibly fraud. |
| Empirical Rule | Understanding how data is spread around the mean | For a roughly Normal feature, values beyond 3 SDs are very rare (about 0.3%), so they deserve a closer look. |

### A few examples in more detail

- **Handling missing data:** imagine 50 people have no BMI recorded. We could fill these gaps with the mean BMI (30.66), because BMI is symmetric. For charges, we would use the median, because the mean is pulled up by a few huge bills.
- **Standard deviation and differences:** if two features have very different SDs, a model that measures "distance" between people (like k-nearest neighbours) would be controlled by the feature with the bigger numbers. Scaling with the mean and SD fixes this.
- **Bayes' Theorem in a model:** a Naive Bayes model learns P(evidence | class) and P(class) from training data. For a new case, it uses Bayes' Theorem to calculate P(class | evidence) and picks the most likely class.
- **Poisson for event counts:** a hospital can model the number of emergency arrivals per hour as Poisson with λ = the average rate, and use it to plan how many doctors are needed.
- **Z-scores for unusual observations:** a factory sensor normally reads 50 °C with SD 2 °C. A reading of 60 °C has z = 5. An anomaly detection system would raise an alert.

**The big idea:** AI/ML models are built on statistics and probability. When you understand these foundations, the models you learn next will make much more sense.
